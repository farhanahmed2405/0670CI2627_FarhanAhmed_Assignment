```python
#1.Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

import pandas as pd
import numpy as np

df1 = pd.DataFrame(np.random.randint(5,15, size=(5, 3)), columns=['A', 'B', 'C'])
df1['Product'] = df1['A'] * df1['B']
print(df1)
```

```python
#2.Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

df2 = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)), columns=['C1', 'C2', 'C3'])
row_sum = df2.sum(axis=1)
print("Sum of rows:\n", row_sum, "\n")
col_sum = df2.sum(axis=0)
print("Sum of coloumns:\n", col_sum, "\n")
print("Original DataFrame:\n", df2)
```

```python
#3.Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

df3 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'], dtype=float)
df3.iloc[1, 0] = np.nan
df3.iloc[3, 1] = np.nan
df3.iloc[2, 2] = np.nan
print("DataFrame with NaNs:\n", df3)
df3_filled = df3.fillna(df3.mean())
print("\n DataFrame after filling NaNs with column means:\n", df3_filled)
```

```python
#4.Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

df4 = pd.DataFrame(np.random.randint(10, 25, size=(6, 4)), columns=['Maths', 'Physics', 'English', 'Hindi'], dtype=float)
df4.iloc[0, 2] = np.nan
df4.iloc[4, 1] = np.nan
print("DataFrame with NaNs:\n", df4, "\n")
df4_dropped = df4.dropna()
print("DataFrame after dropping Nans: \n", df4_dropped)
```

```python
#5 Data Aggregation Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.
import random

categories = [random.choice(['A', 'B', 'C']) for _ in range(10)]
values = np.random.randint(45,85, size=10)
df5 = pd.DataFrame({'Category': categories, 'Value': values})
print("Original DataFrame:\n", df5, "\n")
grouped_df5 = df5.groupby('Category')['Value'].agg(['sum', 'mean'])
print("Aggregated DataFrame:\n", grouped_df5)
```

```python
#6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data.
#Group the DataFrame by 'Category' and compute the total sales for each category.
products = ['Product 1', 'Product 2', 'Product 3', 'Product 4', 'Product 5', 'Product 6', 'Product 7']
categories = [random.choice(['Furniture', 'Kitchen.', 'Cosmetics']) for _ in range(7)]
sales = np.random.randint(230, 450, size=7)
df6 = pd.DataFrame({'Product': products, 'Category': categories, 'Sales': sales})
print("Original DataFrame: \n", df6, "\n")
total_sold = df6.groupby('Category')['Sales'].sum()
print("Total Sales in every Category: \n", total_sold)
```

```python
#7. Merging DataFrames
#Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.
df21a = pd.DataFrame({
    'EmployeeID': [121, 122, 123, 124],
    'Name': ['Ramesh Fernandez', 'Elon Shetty', 'Trump', 'Ram Altman']
})
df21b = pd.DataFrame({
    'EmployeeID': [121, 122, 124, 123],
    'Department': ['Media', 'IT', 'Finance', 'HR']
})
print("DataFrame 1:\n", df21a, "\n")
print("DataFrame 2:\n", df21b, "\n")
merged_df = pd.merge(df21a, df21b, on='EmployeeID')
print("Merged DataFrame:\n", merged_df)
```

```python
#8.MultiIndex DataFrame
#Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame

arrays = [
    ['Group1', 'Group1', 'Group2', 'Group2', 'Group3', 'Group3'],
    ['A', 'B', 'A', 'B', 'A', 'B']
]
multi_index = pd.MultiIndex.from_arrays(arrays, names=('Group', 'section'))
df48 = pd.DataFrame(np.random.randint(1, 20, size=(6, 2)), index=multi_index, columns=['Value 1', 'Value 2'])
print("MultiIndex DataFrame:\n", df48, "\n")
print("Data for Group1:\n", df48.loc['Group1'], "\n")
print("Data for Group1, Subgroup B:\n", df48.loc[('Group1', 'B')], "\n")
print("Slicing DataFrame:\n", df48.loc['Group1':'Group2'])
```

```python
#9.Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.
index = pd.MultiIndex.from_tuples([
    ('Fruit', 'Mango'),
    ('Fruit', 'Dragon Fruit'),
    ('Vegetable', 'Carrot'),
    ('Vegetable', 'Tomato'),
    ('Fruit', 'Mango'),
    ('Vegetable', 'Carrot')
], names=['Category', 'SubCategory'])
df66 = pd.DataFrame({'Values': np.random.randint(17, 67, size=len(index))}, index=index)
print("Original DataFrame:\n", df66, "\n")
sum_df66 = df66.groupby(level=['Category', 'SubCategory']).sum()
print("Sum of Values:\n", sum_df66)
```

```python
#10.Applying Functions Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.
df90 = pd.DataFrame(np.random.randint(5, 42, size=(5, 3)), columns=['P', 'Q', 'R'])
print("Original DataFrame:\n", df90, "\n")
def double_value(x):
    return x * 2
df90_doubled = df90.apply(double_value)
print("Doubled DataFrame:\n", df90_doubled)
```

```python
#11.Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.
df111 = pd.DataFrame(np.random.randint(41, 86, size=(6, 3)), columns=['E', 'F', 'G'])
print("Original DataFrame:\n", df111, "\n")
df111['SumOfColumn'] = df111.apply(lambda row: row['E'] + row['F'] + row['G'], axis=1)
print("DataFrame with Sum :\n", df111)
```

```python
#12.Working with Text Data Create a Pandas Series with 5 random text strings. Convert all the strings to uppercase.
series165 = pd.Series(['Hello', 'World', 'Fox', 'Keyboard', 'Nokia'])
print("Original Series:\n", series165, "\n")
uppercase= series165.str.upper()
print("Uppercase :\n", uppercase)
```
